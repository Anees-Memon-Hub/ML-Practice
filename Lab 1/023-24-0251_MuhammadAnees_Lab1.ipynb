```python
# Q2

def check_prime(number):
    if number < 2:
        return False

    for divisor in range(2, number):
        if number % divisor == 0:
            return False
    return True
print("11:", check_prime(11))
print("14:", check_prime(14))
```

```python
# Q4
def remove_duplicates(values):
    unique_values = []
    for value in values:
        if value not in unique_values:
            unique_values.append(value)
    return unique_values
values = [3, 6, 3, 9, 6, 12, 9, 15]

print("Original list:", values)
print("New list:", remove_duplicates(values))
```

```python
# Q5
def multiply(*values):
    total = 1
    for value in values:
        total *= value
    return total
print("Product:", multiply(3, 4, 2, 5))
```

```python
# Q6
text = "development"
frequency = {
    letter: text.count(letter)
    for letter in set(text)
}
print("Text:", text)
print("Frequency:", frequency)
```
